# 08 · Tilt, phase ramps, and optical masks

A linear phase ramp tilts a beam by shifting its propagation direction. This is the simplest phase mask and a useful way to connect spatial phase to an observable displacement.

This is the package-user version: imports come first, and each operation is introduced when the physical question calls for it.

```python
import numpy as np
import matplotlib.pyplot as plt
from topolino import propTF, gaussian
```

## 1. Apply a linear phase ramp

A plane wave tilted by angle `α` has phase `k x sin(α)`. The envelope is unchanged at the mask plane.

```python
lambda_m = 633e-9
window_m = 8e-3
n = 2048
x_m = np.linspace(-window_m/2, window_m/2, n, endpoint=False)
field0 = gaussian(x_m, 0, 1e-3)
angle_rad = np.deg2rad(0.3)
k_m = 2*np.pi/lambda_m
field_tilted = field0*np.exp(1j*k_m*x_m*np.sin(angle_rad))
```

## 2. Propagate and observe the beam center

A tilted beam walks sideways as it propagates. The small-angle estimate is `x ≈ z tan(α)`.

```python
z_m = 0.30
field_z = propTF(field_tilted, window_m, lambda_m, z_m)
I = np.abs(field_z)**2
center_m = np.sum(x_m*I)/np.sum(I)
print(f"Measured centroid: {center_m*1e3:.3f} mm")
print(f"Small-angle estimate: {z_m*np.tan(angle_rad)*1e3:.3f} mm")
```
